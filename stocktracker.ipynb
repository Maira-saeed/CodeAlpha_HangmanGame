```python
# Task 2: Stock Portfolio Tracker - CodeAlpha Internship

def stock_portfolio():
    # 1. Hardcoded dictionary for stock prices (as per instructions)
    stock_prices = {
        "AAPL": 180,
        "TSLA": 250,
        "GOOGL": 150,
        "MSFT": 400,
        "AMZN": 175
    }

    portfolio = {}
    
    print("--- Welcome to the Stock Portfolio Tracker ---")
    print("Available Stocks and Prices:", stock_prices)
    
    # 2. User Input for stocks and quantity
    while True:
        symbol = input("\nEnter Stock Symbol to add (e.g., AAPL) or 'done' to finish: ").upper()
        if symbol == 'DONE':
            break
        
        if symbol in stock_prices:
            try:
                quantity = int(input(f"How many shares of {symbol} do you own? "))
                portfolio[symbol] = quantity
            except ValueError:
                print("Invalid input. Please enter a number for quantity.")
        else:
            print("Stock not found in our database. Try AAPL, TSLA, GOOGL, MSFT, or AMZN.")

    # 3. Calculate Total Investment
    total_value = 0
    print("\n--- Your Portfolio Summary ---")
    summary_text = "Stock Symbol | Quantity | Price | Total Value\n"
    summary_text += "-" * 45 + "\n"

    for stock, qty in portfolio.items():
        price = stock_prices[stock]
        value = qty * price
        total_value += value
        row = f"{stock:<12} | {qty:<8} | ${price:<5} | ${value}\n"
        summary_text += row
        print(row.strip())

    final_summary = f"\nTotal Investment Value: ${total_value}"
    print("-" * 45)
    print(final_summary)

    # 4. Optional: Save to a .txt file
    save_choice = input("\nDo you want to save this summary to a file? (yes/no): ").lower()
    if save_choice == 'yes':
        with open("portfolio_summary.txt", "w") as f:
            f.write("Stock Portfolio Report\n")
            f.write(summary_text)
            f.write(final_summary)
        print("Summary saved to 'portfolio_summary.txt' successfully!")

if __name__ == "__main__":
    stock_portfolio()
```
